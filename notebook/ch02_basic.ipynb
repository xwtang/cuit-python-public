## 第二章 Python基础语法

### 2.1 变量和对象

在C/C++语言中使用变量a：
```C
int a = 3;
```

在Fortran语言中使用变量a:
```Fortran
integer a
a = 3
```

在Python语言中使用变量a:
```Python
a = 3
```

* 变量a在使用之前不需要声明或者定义
* C和Fortan中的变量a对应固定的内存地址
* Python语言中的变量a仅**指向**整数对象3
* 在程序运行过程中，a可以指向其他对象，没有类型的概念

* 在C/C++和Fortran语言中，在声明变量a之后，仅能改变其数值而不能改变其类型，以下的语句会直接报错：

    C/C++语言:
    ```C
    a = "string";
    ```

    Fortran语言:
    ```Fortran
    a = "string"
    ```

* 而在Python中是完全正确的语法：
```Python
a = 3
a = 'string'
```

* 变量与对象的指向关系
    
    - Python代码中所有数据都是对象，我们通过变量来**指向**(reference)和修改对象
    - 变量(variable)可以指向(reference)**任意**的对象(object)，在代码执行过程中变量的可能操作与其指向的具体对象相关

##### 格式化字符串

格式化字符串是指将数字对象转换为特定格式字符串对象的过程。Python语言提供如下两种转换语法：

* **方法1**: 
```Python
'%d, %f, ...' % (v1, v2, ...)
```
这里的v1，v2为需要格式化的数字，中间`%`符号之前字符串中的`%d,%f`分别对应两个数字的[输出格式](https://docs.python.org/3.7/library/string.html#format-string-syntax)。

````python
'%d,%f' % (1, 2.5)
````

````python
'%5d,%7.2f' % (1, 2.5) 
````

````python
'%05d,%07.2f' % (1, 2.5)
````

* **方法2**：
```Python
'{:d},{:f}'.format(v1, v2)
```
使用字符串对象的`format`方法来格式化字符串。`format`方法的参数为需要格式化的数字，字符串中的`{}`括号对应每个参数的格式。

````python
'{:d},{:f}'.format(1, 2.5) 
````

````python
'{:5d},{:7.2f}'.format(1, 2.5)
````

````python
'{:05d},{:07.2f}'.format(1, 2.5)
````

````python
'{:05d},{}'.format(1, 'Python')
````

* **方法3**：
```Python
f'{v1:d},{v2:f}'
```
使用`f字符`串代替format方法来格式化字符串。format方法的参数放到`f字符串{}`括号中的冒号前面。

````python
a = 1
b = 2.5
f'{a:d},{b:f}'
````

````python
f'{a:d},{b:f}'
````

````python
f'{a:05d},{b:07.2f}'
````

````python
f'{1:05d},{"Python"}'
````

#### 2.3.7 Python文件对象

python的文件对象对应于打开的磁盘文件。文件对象通过内置`open`函数创建：
```Python
fobj = open(filename, mode)
```
filename为文件路径对应的字符串，mode为文件打开方式的字符串。mode的值为如下几种情况：
* 'r'，'w'，'a': 文本模式读取，写入和添加
* 'rb'，'wb'，'ab': 二进制模式下的读取，写入和添加

#### 读取文本文件

文本文件的读写以**行**作为基本单位，常用的两个读取方法如下：
```Python
    line = fobj.readline()
    lines= fobj.readlines()
```

* `readline()`方法每次读取一行，返回字符串对象
* `readlines()`方法一次读取所有行，返回列表对象
* 注意行结束符的平台差异

#### 写入文本文件

文件对象也有两个方法用于文本文件的写入，注意这两个方法参数类型的不同：
```Python
fobj.write(str_obj)
fobj.writelines(list_of_str)
```
* `write()`方法用于写入单行字符串
* `writelines()`用于将一个包含字符串的列表写入文本文件

````python
fout = open('text_out.txt', 'w')

fout.write('write the first line\n')
fout.write('write the second line\n')
fout.write('total number of lines: %d \n' % 2)
fout.writelines(['extra1\n', 'extra2\n'])

fout.close()  
````

````python
fin = open('text_out.txt', 'r')
lines = fin.readlines()
print(lines)
````

````python
print(type(lines), lines)
````

#### 读写二进制文件

* Python语言使用`read()`方法来读取二进制数据，读取的基本单位为**字节**，`read()`方法的原型如下：
```
bytes = fobj.read(N)
```
其中N为需要读取的字节数，返回的bytes为一个字节数组

* Python语言使用`write()`方法来写入二进制数据，写入的基本单位为**字节**，`write()`方法的原型如下：
```
fobj.write(bytes)
```
其中bytes为一个字节数组

* 二进制文件读写复杂性
    - 由于Python语言中最基本的数字都是**对象**，这些对象的内存结构并非传统静态语言那样简单透明
    - 对于读取的bytes数组，按照文件定义的结构使用[struct模块](https://docs.python.org/3.7/library/struct.html?highlight=struct)来将字节数据转化为Python对象
    - 对于很多常用二进制文件格式（比如GRIB, HDF5, NETCDF4等），可以使用扩展库来读取和写入
