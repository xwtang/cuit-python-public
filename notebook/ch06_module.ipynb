## 第六章 Python模块和包

* 模块和包是用于编写大型程序的语言功能，有利于代码功能划分、代码重用、名字空间隔离和系统集成
* 每个**Python源文件**就是一个模块（Module)
* 每个**包含\_\_init\_\_.py的文件夹**就是一个包(Package)
* Python源文件分类
    1. 主程序文件：直接运行的那个.py文件
    2. 模块文件：在主程序文件中通过import加载的.py文件
* 实际大型Python程序的模块和包结构：

### 5.1 模块的创建和使用

#### 5.1.1 创建模块

用文本编辑器编写如下的文件并保存为mod_a.py（注意保存路径）
```Python
# mod_a.py
var_a = 99
L = [1, 'str']

def func_a(x, y):
    print('Hello Python!!!')
    return x + y
```

在mod_a.py相同文件夹下面编辑如下文件，并保存为main.py
```Python
# main.py
import mod_a
mod_a.func_a(1, 2)
print(mod_a.var_a)
```

````python
import mod_a
# mod_a.func_a(3, 5)
print(mod_a.var_a)
````

#### 5.1.2 使用模块

* 使用`import`导入模块完整名字空间
    ```Python
    # main.py
    import mod_a
    print(mod_a.var_a)
    mod_a.func_a(1, 2)
    #mod_a = 'str'
    ```

* 使用`from`加载模块名字空间部分属性
    ```Python
    # main.py
    from mod_a import func_a, var_a
    func_a(1, 2)
    print(var_a)
    ```
* 使用`as`为模块增加**别名**
    ```Python
    import mod_a as mod_a_alias
    from mod_a import func_a as func_a_alias
    
    mod_a_alias.func_a(1, 2)
    func_a_alias(1, 2)
    ```

#### 5.1.3 模块搜索路径

Python的模块文件可以存放在不同的位置，当Python程序执行到导入语句时，将按照下面的规则搜索模块对应的文件：
1. 主程序**根目录**（如果是通过命令行运行程序，即为主程序所在目录；如果是交互式环境，即为当前目录）
2. **PYTHONPATH**环境变量包含的目录
3. 标准库指向的目录(sys.path)
4. Python解释器或者标准库所在目录中.pth文件中包含的目录
5. 第三方模块包目录(site-packages)

````python
import sys
sys.path.append('/usr_package_dir/')
for d in sys.path:
    print(d)
````

#### 5.1.4 模块文件选择

```Python
    import mod
```

Python语言中的模块并非仅以`.py`结尾的文件一种，运行以上代码可能对应于如下几种情况：
1. 一个Python源文件 mod.py
2. 一个Bytecode文件 mod.pyc
3. 一个优化的Bytecode文件 mod.pyo
4. 一个目录名mod的包(package)
5. 一个编译好的扩展模块mod.pyd或mod.so

如果以上的文件在不同的目录被搜索到，那按照搜索顺序加载。当这些文件出现在同一个目录时，不同Python版本可能有不同的操作，所以要尽量避免出现这样的情况。

#### 5.1.5 名字空间与赋值
* 模块和包为**名字空间**对象
* import/from语句导入模块的操作为赋值操作
    ```Python
    from mod_a import var_a, L
    # var_a = mod_a.var_a
    # L = mod_a.L
    var_a = 88
    L[0] = 10
    ```
    以上代码不会修改模块mod_a的var_a，但是会修改L
    ```Python
    import mod_a
    mod_a.var_a = 88
    mod_a.L[0] = 10
    ```
    以上代码会修改模块mod_a的var_a和L


#### 5.1.6 重载模块

* 同一模块在同一Python进程中仅加载一次

* 修改加载后的模块

````python
# 尝试修改mod_a.py文件中func_a()的代码，再次运行这两行程序。
import mod_a
mod_a.func_a(3, 4) 
````

* 使用`reload`函数强制重新加载模块



````python
from importlib import reload
reload(mod_a)
mod_a.func_a(3, 4)
# print(__name__)
````

#### 5.1.7 \_\_name\_\_属性
* 每个Python脚本在运行时都包含一个**\_\_name\_\_**属性
* 脚本作为模块导入时，该属性为模块的名字
* 脚本作为程序直接运行时，该属性为'\_\_main\_\_'

### 5.2 包的创建和使用

* Python的包(Package)是一个包含Python模块文件和特殊`__init__.py`文件的目录
* 包的创建过程仅需创建相应的目录和文件
* 在刚才的创建mod_a.py的目录里面，建立如下的文件和目录结构：
<center><img src="static/package_dir.png" alt="package_dir" style="width: 524px;" /></center>

* 在module1.py中输入如下代码：
    ```Python
    var_module1 = 1
    def func_module1():
        print('module1')
    ```

* 在module2.py中输入如下代码：
    ```Python
    var_module2 = 2
    def func_module2():
        print('module2')
    ```

    这里dir1即为一个Python包，而dir2为子包(sub-package)。module1.py和module2.py是两个不同的模块。

#### 5.2.1 包的导入

* 导入包中的模块同样可以使用`import`和`from`两种语法
```Python
import dir1.module1
import dir1.dir2.module2
```

* 使用这种方式导入的模块，在使用时需要输入完整的包路径
```Python
dir1.module1.func_module1()
dir1.dir2.module2.func_module2()
```

````python
import dir1.module1
import dir1.dir2.module3
````

````python
dir1.module1.func_module1()
dir1.dir2.module2.func_module2()
````

````python
module1.func_module1() # 错误，函数名不存在
````

* 为了方便使用包中的模块，通常我们使用`from`语句来导入包中的模块和名字，例如：
```Python
from dir1 import module1
from dir1.dir2 import module2
```
    或者
```Python
from dir1.module1 import func_module1
from dir1.dir2.module2 import func_module2
```

````python
from dir1 import module1 
from dir1.dir2 import module2
````

````python
module1.func_module1()
module2.var_module2
````

````python
from dir1.module1 import func_module1
from dir1.dir2.module2 import func_module2
````

````python
func_module1()
func_module2()
````

## 常用的标准库模块
* `os`：包含与操作系统相关的功能
    * 目录遍历，查询文件状态
    * 创建删除文件
    * 构建与修改文件路径
* `sys`：操作系统和Python运行环境信息
* `re`：正则表达式，进行复杂的字符串匹配
* `json`：json数据读写
    * json.load 加载json文件
    * json.dump 将json文件存为文件
* `math`：数学函数库
