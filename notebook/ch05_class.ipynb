## 第6章 Python类（class）

Python程序的执行过程可以比喻为“使用东西(对象)完成事情(do things with stuff)”，**对象**是Python语言中最重要的概念。
* 东西（stuff）就是Python语言里面的对象(object)
* 事情就是对象的属性、方法和内置函数

使用Python类可以创建用户自定义对象，从而支持完整的面向对象编程

### 6.1 面向对象编程基本概念

### 6.2 Python相关概念

* 两种OOP相关对象
    * 类对象：包含公用方法（函数）
    * 实例对象：包含被处理的数据
* 关联名字空间
    * 类和实例对象都是名字空间
    * 实例与类（子类与父类）的名字空间相互关联

* 类方法（函数）
    * 类内部创建的函数自动获得一个参数`self`
    * 类可以创建特殊函数以捕捉内置操作符


### 6.3 Python类编程基础
#### 创建类与实例

* 语法要点
    - `class`语句创建类对象并指定名字
    - `class`语句内的赋值语句创建新的类属性
    - 类对象为实例对象提供公用的功能

* 实例对象
    - 将`class`按函数的方式调用产生实例对象
    - 对象获得类属性和独立名字空间
    - 在类方法中`self`变量的属性为实例独有

````python
class FirstClass:
    def setdata(self, value):
        self.data = value
    def display(self):
        print(self.data) 
````

````python
X = FirstClass()
Y = FirstClass()
````

#### 继承
子类获得父类的全部功能，但不影响父类本身（通过定义新的子类来修改或者增加功能），语法要点：
* 创建新类时使用括号列出父类
* 新类自动获得所有父类的属性（继承）
* 实例对象获得所有相关类的属性
* 实例对象的属性操作触发“属性搜索”

使用子类修改父类的功能，而不是直接修改父类代码

```Python
class FirstClass:
    def setdata(self, value):
        self.data = value
    def display(self):
        print(self.data)
```

````python
class SecondClass(FirstClass):
    def display(self):
        print('value = "{}"'.format(self.data))
````

````python
Z = SecondClass()
````

````python
Z.setdata(42)
Z.display()
````

```Python
X = FirstClass()
```

````python
X + Y
````

#### 操作符重载

```Python
print(1 + 2)
print('Python' + 'Awesome')
```

* 不同的对象类型都支持相同的“加法”操作
* 操作符重载：自定义系统包含的操作符，如：(`+，-，*，/`)
* 用户创建的实例对象的“行为”与核心对象一致

* 语法要点
    - 双下划线开头和结尾的特殊函数，如:\_\_init__()
    - 实例对象参与内置操作符（+，-，*，/，...）运算时，自动调用对应的特殊函数
    - 用户可重载绝大部分操作符
    - 操作符重载是可选功能

````python
class ThirdClass(SecondClass):
    def __init__(self, value):
        self.data = value
    def __add__(self, other):
        return ThirdClass(self.data + other)
    def __str__(self):
        return f'[string: {self.data}]'
````

* 创建了三个特殊函数：
   - \_\_init\_\_在创建实例时被调用
   - \_\_add\_\_在加法操作（+）是被调用
   - \_\_str\_\_在字符输出（print）的时候被调用

````python
print(X)
````

````python
a = ThirdClass('abc')
````

````python
a.display()
````

````python
print(a)  # a.__str__()
````

````python
b = a + 'xyz'
````

````python
b.display()
````

````python
print(b)
````

### 6.3 异常处理

#### 异常简介

* 表示程序执行期间发生错误的特殊Python对象
* 编写程序时**捕获**特定异常事件，程序将继续运行；如果未对异常进行处理，程序将终止并显示一个traceback
* 除了捕获系统生成的异常，用户代码也可以主动**抛出**异常


#### 异常类

* 继承自相同的父类BaseException
* 捕获父类对象将捕获相应的子类型

<center><img src=static/exception_class.png></center>

#### 捕获异常

```Python
try:
    # 用户代码
except error1:
    # 出现异常执行
except error2:
    # 出现异常执行
else:
    # 未出现异常执行
finally:
    # 是否出现异常都执行
```

* try,except语句为复合语句
* try只能出现一次，except可出现多次
* 按顺序匹配except语句指定的类型

````python
l = [1, 2, 3]
d = {'k1': 0.0, 'k3':1.0}
# l[3] / d['k2']
try:
    print(l[2], d['k3'])
    l[1] / d['k1']
except KeyError:
    print('key error!')
except IndexError:
    print('index error!')
except ZeroDivisionError:
    print('divided by zero !!!')
````

#### 抛出异常

```Python
raise 异常类
```

````python
l = [1, 2, 3]
idx = 4
try:
    if idx < 0 or idx > len(l):
        raise KeyError()
    print(f'the second item of list:', l[idx])   
except TypeError:
    print('类型错误')
````
