### 4.1 创建和调用函数
```Python
def func_name(arg1, arg2,... argN):
    statements
    return value
```

* 创建函数对象使用`def`语句。没有**定义**函数这样的概念，函数在语句执行时被动态创建
* `def`语句是一个复合语句，代码需要缩进
```Python
if test:
    def func(x):
        x += 1
        return x 
else:
    def func(x):
        x -= 1
        return x
func(1) 
```

#### 函数名和函数对象

````python
def sum(x, y):
    z = x + y
    return z
````

````python
sum(1, 2)
````

````python
sum_alias = sum
sum_alias(3, 4)
````

````python
def times(x, y):
    return x * y
times(3, 4)
````

````python
# 函数名sum重新指向函数名times对应的函数对象
sum = times
sum(1, 2)
````

````python
# 函数名指向字符串，不能再被调用
sum = 'str'
sum(1, 2)
````

#### Python名字作用域

* 名字作用域：Python代码中名字的有效范围（生存期）
    * 全局域(global)：单个python源程序文件
    * 局部域(local)：单个函数内部
    * 包围域(Enclosing)：嵌套函数的外部函数
    
* LEGB名字查找规则:
    * 首先搜索局部域(Local)
    * 其次搜索包围域(Enclosing)
    * 再次搜索全局域(Global)
    * 最后搜索内置域(Build-in)
    
* 如果在以上4个域中无法查找到名字，程序将报错。

````python
X = 99
def func():
    X = 88
    print('X in func:', X)
func()
print('X in global:', X)
````

#### 跨作用域访问

* 名字的作用域阻止了在局部域中修改全局域中的名字
* 可以使用`global`和`nonlocal`语句进行跨作用域访问
    * `global` 在局部域中修改全局域变量名
    * `nonlocal` 在局部域中修改包围域中的变量名

#### 例子一：局部域修改全局域

````python
X = 99
def func():
    X = 88
func()
print(X)
````

````python
X = 99
def func():
    global X
    X = 88
func()
print(X)
````

#### 例子二：局部域修改包围域

````python
X = 99
def f1():
    X = 88
    def f2():
        X = 77
    f2()
    print('X in f1:', X)
    
f1()
print('X in global:', X)
````

````python
X = 99
def f1():
    X = 88
    def f2():
        nonlocal X
        X = 77
    f2()
    print('X in f1:', X)
f1()
print('X in global:', X)
````

### 4.3 函数参数传递

* `def`语句的函数参数列表成为函数体的局部域名字
* 参数传递是对这些局部域名字**赋值**的过程
    ```Python
    def func(a):
        #a = 8
        a = a + 1
        return a
    func(8)
    ```
    
    在调用func时，隐含了```a = 8```这个语句

* 在函数中修改**可修改类型**的对象可能影响函数调用者
    * **不可修改immutable**类型参数可等同为“传值”
    * **可修改mutable**类型参数可等同为“传址”

````python
def func(a):
    a = 99

b = 88
func(b)
print(b)
````

````python
def changer(a, b):
    a = 2
    b[0] = 'str'

X = 1
L = [1, 2, 3]

print(X, L)
changer(X, tuple(L)) # tuple(L)
print(X, L)
````

### 4.4 Python函数的参数匹配

```Python
def sum(x, y):
    return x + y
sum(3, 4)
```

```Python
x = 3
y = 4
```

* 传入参数个数与创建函数时的参数个数相同
* 传入参数与创建函数时的参数顺序一一对应

* 在创建Python函数对象时，函数的参数列表可以有如下4种形式:
    - 普通参数
       ```Python
       def func(name)
       ```
    - 默认值参数
       ```Python
       def func(name=value)
       ```
    - 任意位置参数
       ```Python
       def func(*name)
       ```
    - 任意关键字参数
       ```Python
       def func(**name)         
       ```
* 四种参数列表同时存在，需按如下的顺序排列：
   ```Python
   def func(name, default=value, *pos, **keyword)
   ```

* **调用**Python函数时，参数列表也可以有如下4种形式：
    - 位置参数，按位置匹配
       ```Python
       func(obj)
       ```
    - 关键字参数，按名字匹配
       ```Python
       func(name=obj)
       ```
    - 序列展开，作为多个位置参数传递
       ```Python
       func(*obj)
       ```
    - 字典展开，作为多个关键字参数传递
       ```Python
       func(**obj)
       ```

* 这四种参数同时存在时，需要按如下的顺序排列：
   ```Python
   func(name, keyword=value, *seq, **dicts)
   ```

#### 函数创建时仅包含普通参数

````python
def func(a, b, c):
    print(f'a={a}, b={b}, c={c}')
````

调用函数时的各种情况

````python
func(1, 5, 3)       # 按位置匹
````

````python
func(b=5, c=3, a=1) # 按关键字匹配
````

````python
args = (1,5,3)
func(*args)         # 展开可迭代对象，并按位置赋值 a, b, c = args
````

````python
args = {'c':3, 'a':1, 'b':5}
func(**args)        # 展开字典，并按关键字匹配 
````

### 4.5 lambda函数

* 匿名函数: 创建之后不赋值给函数名，而是直接返回函数对象。

    lambda函数的创建使用如下的表达式：
    ```Python
    lambda arg1, arg2,... argN : expression
    ```

* 在使用lambda函数时需要注意以下两点：
    - lambda只是一般表达式，不是一个语句
    - lambda的函数体只能是单个表达式

````python
def func(x, y, z):
    return x + y + z
func(1, 2, 4)
````

````python
func = lambda x, y, z : x + y + z
func(1, 2, 4)
````

* 函数的参数之一为函数对象，如functional编程：
    - `map(func, *iterable)`
    - `filter(func, *iterable)`
    - `reduce(func, *iterable)`

````python
L = [5, 7, 4, 2]
L1 = [x + 10 for x in L]
print(L1) 
````

````python
list(map(lambda v: v+10, L))
````

````python
L2 = [x + 10 for x in L if x > 3]
print(L2) 
````

````python
# list(filter(lambda v: v>3, L))
list(map(lambda x: x + 10, filter(lambda v: v > 3, L)))
````

````python
list(map(lambda x: x[0] + x[1], zip(L, L1)))
````

````python
def add(x):
    return x[0] + x[1]
list(map(add, zip(L, L1)))
````

### 4.6 函数修饰器

* 装饰器(decorator)：用于装饰其他Python对象的语法结构，可以在不修改原有代码的情况下，改变或增加被装饰对象的功能

* Python装饰器是一个可**调用（callable）对象**，被调用时返回另一个**可调用对象**

* Python装饰器主要作用于函数和类，以@操作符开始，出现在函数和类的创建代码之前

````python
def do_nothing(f):
    print("running a Python decorator ...")
    return f
````

````python
do_nothing()
````

````python
@do_nothing
def func():
    print('func ...')
````

````python
func()
````

#### 函数装饰器实例

````python
def outer(func):
    def inner():
        print('内部操作！')
        result = func()
        print('操作完毕！')
        return result
    return inner

@outer
def f1():
    print('操作开始')
````

````python
f1()
````
