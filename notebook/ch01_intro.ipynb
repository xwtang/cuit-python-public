## 第一章 Python概述

### 1.1 Python发展历史

* <img style="float: right; width: 300px; margin: 5%" src="static/Guido.jpg">
Python语言是由荷兰数学家吉多·范罗苏姆（Guido van Rossum）于1989年开发的一种新脚本语言 
* 名字Python取自BBC电视剧——Monty Python's Flying Circus
* 吸取了之前设计程序语言失败的教训，Guido决心要将Python设计成为一门“优雅”，“简洁”的编程语言





* Python语言目前主要包含两个大的版本分支2.x和3.x
* 2.x系列中2.6和2.7两个版本仍被使用，而3.x系列已发展至3.12版
* 2.x系列官方已停止功能更新，推荐使用3.x（3.11以上）版本
* 2.x和3.x版本存在部分不兼容的语法，使用他人程序时需要注意

### 1.2 Python语言特征和应用领域

* **动态性**：**变量名**都没有类型的概念，只有**对象**有类型的概念
* **解释性**：代码无需编译，由Python解释器(PVM)直接运行
* **交互性**：可以逐句输入程序代码并实时查看运行结果
* **多种编程范式**：面向过程、面向对象、面向函数

#### 1.2.1 Python与其他脚本语言（bash，awk，sed）

* 脚本语言(Script Language)：用于和操作系统交互的小程序，例如批量处理文件和启动其他应用程序，管理系统资源等
* Python语言可以实现与传统脚本语言类似的功能（使用标准库的os，sys模块）
* Python具备开发大型软件项目的特性（类，模块和模块包）

#### 1.2.2 Python与静态语言（Fortran，C/C++)

* 解释运行：Python程序无需编译，编写完成即可运行；交互模式下，可同时**编写并运行**Python程序。
* 动态类型：Python语言的变量没有类型的概念，可以**指向**任意类型的对象；静态语言的所有变量都需要预先定义类型，且在使用过程中需保证变量类型的一致性。

#### 1.2.3 Python程序的组成与运行方式

<center><img src="static/python_component.png" alt="runtime" style="width:600px"/></center>

* Python解释器: 语言本身仅定义了一系列的数据对象、语法和程序结构，有多种具体的实现方式：
    - CPython: C语言实现
    - Jython: Java语言实现
    - IronPython: Windows .net 实现
    - Pypy: Python语言实现（即时编译）

* Python程序的执行分为如下两个阶段：
    1. 源代码 --> 字节代码（Bytecode）
    2. 字节代码 --> CPU指令
    <img src="static/python_runtime.png" alt="runtime" style="width:512px" />
* 字节代码由Python虚拟机转换为相应的CPU指令

#### 1.2.4 Python的应用领域

* Python扩展库：Python语言的强大功能主要通过扩展库实现
    - Python语言本身包含了一个**标准库**
    - 而特定领域相关的功能则是通过第三方库来实现的
* 基于这些扩展库，Python主要用于如下方面的应用开发
    - 系统脚本程序（os, sys）
    - 桌面应用程序（QT, Wxwidget, MFC）
    - 网络应用程序（socket, urllib）
    - 数据库开发 (sqlalchemy)
    - 数值和科学计算（numpy, scipy, matplotlib, pandas）
    - 人工智能开发与应用（pytorch, tensorflow）

### 1.5 运行第一个Python程序

Python源程序: **script1.py**
```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0., 2*np.pi, 361)
y = np.sin(x)

plt.plot(x, y)
plt.show()

```

#### 1.5.1 运行Python程序的两种方式

* **交互运行**
    1. 启动**Python命令行**
    2. 输入以上代码

* **命令行运行**
    1. 启动Anaconda Prompt命令行环境
    2. 编辑代码并保存为script1.py
    3. 输入 python script1.py

* **IDE运行**
    1. 创建项目
    2. 配置项目
    3. 编辑以上文件并运行
