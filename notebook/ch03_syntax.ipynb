### 3.3 选择（if）语句

程序运行过程中通常需要根据一定的条件来执行不同的操作，这就需要用到`if`语句。`if`语句是一种复合语句，它的一般形式：
```Python
if test1:
    statement block1
elif test2:
    statement block2
else:
    statement block3
```
其中：
   * elif和else是可选的
   * elif可以出现多次
   * else只能出现一次
由于`if`是复合语句，在书写代码时需要注意代码缩进的问题。

````python
if 2 > 1:
    print('True')
````

````python
x = 11
if x > 10:
    print('x>10')
elif x > 6:
    print('x between 6 and 10')
else:
    print('x<6')
````

在复合语句中嵌套复合语句时，缩进的原则依然适用。建议所有程序保持一致的缩进（如2或4个空格）。

````python
x = 6
if x > 10:
    print('x>10')
elif x > 5:
    if x == 6:
        print('x==6')                 # 注意这里需要再次缩进
        a = 1
    else:
        print('x between 7 and 10')
        a = 0
else:
    print('x<6')
````

### 3.4 循环语句

* Python语言包含`while`和`for`两种循环语句
* `while`循环和静态语言中的`while`循环语法结构一致，主要用于基于某种条件重复执行一个代码块
* `for`循环主要用于顺序获取一个序列的元素，和C/C++语言中的`for`循环有较大的区别

#### 3.5.1 while循环

`while`语句是一种复合语句，它的一般形式为：
```Python
while test:
    statement
    if test1: break
    if test2: continue
else:
    statement
```
`else`语句在非break结束循环的条件下执行

````python
x = 1
while x < 10:
    print(x, end=',')
    x += 1
    if x > 6:
        print('break')
        break
else:
    print('else')
````

#### while循环中的**else**

当循环体在非break状态下退出时，else语句内的代码块会被执行

* 字符串查找代码（Ver.1）

````python
x = 'Merry Xmas'

flag = False
while x and not flag:
    if x[0] == 'X':
        flag = True
        break
    else:
        x = x[1:]

if not flag:
    print('not found X!')
else:
    print('found X!')
````

* 字符串查找代码（Ver.2）

````python
x = 'Merry xmas'
while x:
    if x[0] == 'X':
        print('found X!')
        break
    else:
        x = x[1:]
else:
    print('not found X!')
````

#### 3.5.2 continue 和 break

* `break`语句用于跳出循环体并结束当前循环
* `continue`语句用于跳过当前循环`continue`语句后面的代码，并继续下次循环

````python
while True:
    x = input()
    if int(x) < 0:
        break
    else:
        print('your input: ' + x)
````

````python
x = 0
while x < 10:
    x += 1
    if x % 2 == 0: continue
    print(x, end=' ')
````

#### 3.5.3 for循环结构

`for`语句是一种复合语句，它的一般形式为：

```Python
for v in object:
    statement
    if test1: break
    if test2: continue
else:
    statement
```
* object为序列（字符串，列表和元组）或可迭代（iterable）对象（包括字典和文本文件）
* object为字典时，序列为字典的**键**
* 名字v称为**循环变量**，在循环中依次被**赋值**为序列的一个元素

#### for循环实例

````python
for v in 'Python':
    print(f'v={v}', end=', ')
````

````python
for v in (1, 2, 'a'):
    print(f'v={v}', end=', ')
````

````python
D = {'name': 'China', 'year':1949}
for k, v in D.items():
    print(k, v) 
````

#### for循环序列展开赋值

for循环在每步循环开始前都隐含了一个赋值语句，赋值语句中的基本赋值和序列展开赋值都适用

````python
L = [(1, 2), (3, 4), (5, 6)]
for a, b in L: # a, b = (1, 2)
    print(a, b)
````

````python
for k, v in D.items():
    print(k, '<-->', v)
````

#### 传统的计数循环
Python语言使用`range()`函数创建整型序列，可模拟C/C++语言中的`for`循环和Fortran语言中的`do`循环。

`range()`函数的原型如下：
```Python
range(N1,N2,NI)
```
其中N1表示开始，N2表示结束，NI表示增量。

````python
for i in range(5, 12):
    print(i)
````

````python
for i in range(2, 10):
    print(i)
````

````python
for i in range(10, 2, -2):
    print(i)
````

下面的两个C/C++和Fortran语言中的循环语句：

C/C++语言
```C
for(int i=0; i<10; i+=2){
    printf("%d", i);
}
```

Fortran语言
```Fortran
do i=0,9,2
    print*, i
end do
```
可以用下面的Python语言`for`循环代替：

````python
for i in range(0, 10, 2):
    print(i)
````

* `enumerate`函数：同时获取元素的位置和值

````python
i = 0
for v in L:
    print(i, '<-->', v)
    i += 1
````

````python
for i, v in enumerate(L):
    print(i, '<-->', v)
````

* `zip`函数：循环多个等长的序列

````python
L1 = [1, 2, 3]
L2 = [4, 5, 6]
for v1, v2 in zip(L1, L2): #[(1,4), (2,5), (3,6)]
    print(v1, v2)
````

### 3.5 列表和字典的递推构造(comprehension)

* 创建或者修改列表和字典对象时，可使用循环对每个元素进行操作（推导式语法）。例如如下的代码：

````python
L = [5, 7, 4, 2]
for i in range(4):
    L[i] += 10
print(L)
````

* 上面代码的功能可通过以下递推构造实现

````python
L = [5, 7, 4, 2]
L = [x + 10 for x in L]
print(L) 
````

列表递推构造的一般形式：
```Python
L = [ expression for v in object if test ] 
```

递推构造只是一种更加简洁的语句，其功能可以用传统的循环代替。

````python
Ls = ['3.12', '5.45', '4.5']
L = [f(v) for v, f in zip(Ls, [float,float,float])]
print(L)
````

````python
L = [5, 7, 4, 2]
L1 = [x + 10 for x in L if x > 3]
print(L1)
````

#### 字典的递推构造

字典递推构造的一般形式：
```Python
D = { key(v):value(v) for v in object if test } 
```

````python
L = [5, 7, 4, 2]
D = {f'item:{v}' : v**2 for v in L}
print(D)
````

````python
D = {i:v for i,v in enumerate(L)}
print(D)
````

````python
L = [5, 7, 4, 2]
L1 = ['a', 'b', 'c', 'd']
D = {k:v for k, v in zip(L, L1) if k != 4}
print(D)
````
